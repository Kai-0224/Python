```python
def backtrack(path, choices):
    if base_case_condition:  # If a valid solution is found
        store_solution(path)
        return
    
    for choice in choices:
        if is_valid(choice):  # Prune invalid choices (optional)
            make_choice(path, choice)  # Modify the path
            backtrack(path, new_choices)  # Recur with the updated state
            undo_choice(path, choice)  # Undo the choice (backtrack)
```

```python
### Letter combinations of a phone number
class Solution(object):
    def letterCombinations(self, digits):
        """
        :type digits: str
        :rtype: List[str]
        """
        if not digits: # No numbers
            return []
        phone = {"2": "abc", "3": "def", "4": "ghi", "5": "jkl", "6": "mno", "7": "pqrs", "8": "tuv", "9": "wxyz"}
        sol = []
        def backtrack(combi, next_digi):
            if not next_digi:
                sol.append(combi) 
                return # No number -> stop
            for letter in phone[next_digi[0]]: 
                backtrack(combi + letter, next_digi[1:])
        backtrack("", digits)
        return sol
```

```python
### Generating All Subsets
def subsets(nums):
    result = []
    
    def backtrack(path, index):
        result.append(list(path))  # Store a copy of the current subset
        
        for i in range(index, len(nums)):  
            path.append(nums[i])  # Choose an element
            backtrack(path, i + 1)  # Recur with the next index
            path.pop()  # Undo the choice (backtrack)
    
    backtrack([], 0)
    return result

print(subsets([1, 2, 3]))
```

```python
### Solving N-Queens Problem
def solveNQueens(n):
    solutions = []
    board = [["."] * n for _ in range(n)]
    
    def is_safe(row, col):
        for i in range(row):
            if board[i][col] == "Q" or \
               (col - (row - i) >= 0 and board[i][col - (row - i)] == "Q") or \
               (col + (row - i) < n and board[i][col + (row - i)] == "Q"):
                return False
        return True
    
    def backtrack(row):
        if row == n:
            solutions.append(["".join(r) for r in board])
            return
        
        for col in range(n):
            if is_safe(row, col):
                board[row][col] = "Q"  # Place Queen
                backtrack(row + 1)  # Move to the next row
                board[row][col] = "."  # Remove Queen (backtrack)
    
    backtrack(0)
    return solutions

print(solveNQueens(4))
```
