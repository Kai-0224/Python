```python
l1 = [2, 4, 3]
l2 = [5, 6, 4]
```

```python
class Solution(object):
    def addTwoNumbers(self, l1, l2):
        """
        :type l1: Optional[ListNode]
        :type l2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        l1.reverse()
        l2.reverse()
        s1 = ''
        s2 = ''
        for i in l1:
            s1 += str(i)
        for j in l2:
            s2 += str(j)
        tem_add = int(s1) + int(s2)
        s3 = str(tem_add)
        l3 = []
        for k in s3:
            l3.append(int(k))
        l3.reverse()
        return l3
```

```python
solution = Solution().addTwoNumbers(l1, l2)
```

```python
def convert(s, numRows):
    """
    :type s: str
    :type numRows: int
    :rtype: str
    """
    rows = [''] * numRows #Create a matrix
    k = 0
    direction = (numRows == 1) - 1 #Define the direction to add characters
    for c in s:
        rows[k] += c
        if k == 0 or k == numRows - 1:
            direction *= -1 #Change the direction for the head/tail of matrix
        k += direction
    return ''.join(rows)
```
