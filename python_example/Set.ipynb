```python
set1 = set()# 建立空的集合

set2 = {1, 2, 3, 4, 5}

# 從串列 (List) 來建立集合
set3 = set([i for i in range(20, 30)])

# 從數組 (Tuple) 來建立集合
set4 = set((10, 20, 30, 40, 50))

# 集合不會包含重複的資料, 你可以從 set5 印出來的結果得知
set5 = set((1, 1, 1, 2, 2, 3))

print(set1)
print(set2)
print(set3)
print(set4)#(order may vary)
print(set5)
```

```python
set2 = {1, 2, 3, 4, 5}
set4 = set((10, 20, 30, 40, 50))

set2.add(6) # 在 set2 中加入 6
set4.remove(20) # 將 set4 中的 20 刪除

print(set2)
print(set4)
```

```python
setA = {1, 6, 8, 10, 20}
setB = {1, 3, 8, 10}

# 以下兩個都是 聯集 的寫法
print(setA.union(setB))
print(setA | setB)

# 以下兩個都是 交集 的寫法
print(setA.intersection(setB))
print(setA & setB)

# 以下兩個都是 差集 的寫法
print(setA.difference(setB)) #in A but not in B
print(setA - setB)
print(setB.difference(setA)) #in B but not in A
print(setB - setA)

# 以下兩個都是 對稱差集 的寫法
print(setA.symmetric_difference(setB))#in either A or B, but not in both
print(setA ^ setB)
```

```python
setA = {1, 6, 8}
setB = {1, 6, 8, 10, 20}

# 使用 issubset() 來判斷 A 是否為 B 的子集合
print(setA.issubset(setB))

# 使用 issuperset() 來判斷 B 是否為 A 的子集合
print(setB.issuperset(setA))
```

```python
setA = {1, 6, 8}
setB = {1, 6, 8, 10, 20}

print(setA == setB)
print(setA != setB)
```
