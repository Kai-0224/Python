```python
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val # Stores the value of the node
        self.next = next # Points to the next node (or None if it's the last node)
```

```python
def list_to_linked_list(lst):
    head = ListNode(lst[0])
    current = head
    for val in lst[1:]:
        current.next = ListNode(val)
        current = current.next
    return head
```

```python
def linked_list_to_list(node):
    result = []
    while node:
        result.append(node.val)
        node = node.next
    return result
```

```python
class Solution(object):
    def addTwoNumbers(self, l1, l2):
        """
        :type l1: Optional[ListNode]
        :type l2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        def toint(node):
            return node.val + 10 * toint(node.next) if node else 0
        def tolist(n):
            node = ListNode(n % 10)
            if n > 9:
                node.next = tolist(n // 10)
            return node
        return tolist(toint(l1) + toint(l2))
```

```python
# Convert lists to linked lists
l1 = list_to_linked_list(a)
l2 = list_to_linked_list(b)

# Create Solution instance
solution = Solution()

# Add two numbers
result = solution.addTwoNumbers(l1, l2)

# Convert the result back to a list and print
print(linked_list_to_list(result))
```
