```python
import pandas as pd
import numpy as np
from sklearn import datasets
```

```python
# 1. Dataset
ds = datasets.load_breast_cancer()
X =pd.DataFrame(ds.data, columns=ds.feature_names)
y = ds.target
```

```python
# 2. Data clean
# no need
```

```python
# 3. Feature Engineering
# no need
```

```python
# 4. Split
from sklearn.model_selection import train_test_split    
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
# 5. Define and train the KNN model
from sklearn.neighbors import KNeighborsClassifier

clf = KNeighborsClassifier(n_neighbors = 3)

# 適配(訓練)，迴歸/分類/降維...皆用 fit(x_train, y_train)
clf.fit(X_train, y_train)

# algorithm.score: 使用 test 資料 input，並根據結果評分
print(f'score={clf.score(X_test, y_test)}')
```

```python
# 驗證答案
print(' '.join(y_test.astype(str)))
print(' '.join(clf.predict(X_test).astype(str)))
```

```python
# 查看預測的機率
print(clf.predict_proba(X_test.head()))
```

```python
from sklearn.metrics import plot_confusion_matrix
import warnings
#suppress warnings
warnings.filterwarnings('ignore')
plot_confusion_matrix(clf, X_test, y_test)
```
