# 1. Tensor 張量

### 1-1. Vector 向量

```python
import numpy as np
import matplotlib.pyplot as plt

# We'll use a numpy array for our vector
v = np.array([2,1])
origin = [0], [0]

plt.axis('equal')
plt.grid()
plt.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
plt.quiver(*origin, *v, scale=10, color='r')    # quiver畫箭頭，*表示展開

plt.show()
```

### 1-2. 使用 numpy 的 norm 函數計算距離

```python
import numpy as np

vMag = np.linalg.norm(v)
print (vMag)
```

### 1-3. Calculating Direction 使用 atan 函數計算方向

```python
import math
import numpy as np

# 向量座標
v = np.array([2,1])
vTan = v[1] / v[0]
print ('tan = ' + str(vTan))

# 三角函數轉換
vAtan = math.atan(vTan)
print('radian =', vAtan)                   # 逕度

# 或 math.degrees(vAtan)
print('degree =', vAtan*(360/(2*math.pi))) # 角度
```

### 1-4. Superposition

```python
import math
import numpy as np
import matplotlib.pyplot as plt

v = np.array([2, 1])    # 力 1
s = np.array([-3, 2])   # 力 2
z = v + s               # 合力

# Plot v and s
vecs = np.array([v, s, z])
origin = [0, 0, 0], [0, 0, 0]
plt.axis('equal')
plt.grid()
plt.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
plt.quiver(*origin, vecs[:, 0], vecs[:, 1], color=['r', 'b', 'g'], scale=10)

plt.show()
```

# 2. Vector Multiplication 向量相乘

### 2-1. Scalar Multiplication 純量乘法

```python
import numpy as np
import matplotlib.pyplot as plt
import math

v = np.array([2,1])
w = 2*v

# 作圖
vecs = np.array([w, v])
origin = [0, 0], [0, 0]
plt.grid()
plt.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
plt.quiver(*origin, vecs[:, 0], vecs[:, 1], color = ['k', 'r'], scale=10)

plt.show()
```

### 2-2. Inner product 內積

```python
import numpy as np

v = np.array([2, 1])
s = np.array([-3, 2])

d = v @ s
# 或 d = v.dot(s)
print(d)
```

```python
import math

vMag = np.linalg.norm(v)
sMag = np.linalg.norm(s)

cos = v @ s / (vMag * sMag)
theta = math.degrees(np.arccos(cos))

print(theta)
```

### 3-2. Matrix Operations

```python
import numpy as np
A = np.array([[1,2,3],
              [4,5,6]])
B = np.array([[6,5,4],
              [3,2,1]])
C = np.array([[9,8],
              [7,6],
              [5,4]])
```

```python
print(A + B)
```

#### B. 矩陣減法

```python
print(A - B)
```

#### C. 矩陣乘法

#### C1. Scalar Multiplication

```python
print(2 * A)
```

#### C2. Dot

```python
print(A @ C) # 矩陣 2*3 x 矩陣 3*2 → 矩陣 2*2
```

```python
print(C @ A) # 矩陣 3*2 x 矩陣 2*3 → 矩陣 3*3
```

### 3-4. Inverse of a Matrix 反矩陣

```python
import numpy as np
a = np.array([[1, 2],
              [3, 4]
             ])
print(np.linalg.inv(a))
```

### 可以利用反矩陣特性，來求方程式的解

$x + y = 16$<br>
$10x + 25y = 250$

```python
A = np.array([
    [1, 1],
    [10, 25]])
B = np.array([16, 250])
print(np.linalg.inv(A) @ B)
# 其結果會等同 np.linalg.solve(A, B)
```

# 4. Transformation, Eigenvector & Eigenvalue 轉換, 特徵向量與特徵值

### 4-1. Linear Transformations 線性轉換

#### A. 轉換距離，方向不變

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([1, 0])
A = np.array([[2,1],
              [0,2]])

t = A @ v

original = [0], [0]
plt.axis('equal')
plt.grid()
plt.ticklabel_format(style='sci', axis='both', scilimits=(0, 0))
plt.quiver(*origin, *t, color='orange', scale=10)
plt.quiver(*origin, *v, color='blue', scale=10)
plt.show()
```

#### B. 轉換方向，距離不變

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([1,0])
A = np.array([[0,-1],
              [1,0]])

t = A@v
print (t)

origin = [0], [0]
plt.axis('equal')
plt.grid()
plt.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
plt.quiver(*origin, *t, color='orange', scale=10)
plt.quiver(*origin, *v, color='blue', scale=10)

plt.show()
```

#### C. 方向、距離均改變

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([1,0])
A = np.array([[2,1],
              [1,2]])

t = A@v
print (t)

origin = [0], [0]
plt.axis('equal')
plt.grid()
plt.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
plt.quiver(*origin, *t, color='orange', scale=10)
plt.quiver(*origin, *v, color='blue', scale=10)

plt.show()
```

### 4-2. 特徵向量與特徵值 (Eigenvectors and Eigenvalues)

```python
v = np.array([1,0])
A = np.array([[2,0],
              [0,2]])

t1 = A@v
t2 = 2*v

print(t1, t2)
```

```python
import numpy as np
A = np.array([[2, 0],
              [0, 3]])
eVals, eVecs = np.linalg.eig(A)
print('Eigenvalue:\n', eVals)
print('Eigenvector:\n', eVecs)
```
